  # Question Three: Missing Number in Range
 
  You are given a list containing `n` integers in the range `[0, n]`. Return a list of numbers that are missing from the range `[0, n]` of the array. If there is no missing number, return -1. Note, all the integers in the list may not be unique.
  
  ## Examples

  ### Example 1

  Input: `lst = [0, 2]`

  Output: [1]

  ### Example 2

  Input: `lst = [5, 0, 1]`

  Output: [2, 3, 4]

  ### Example 3

  Input: `lst = [6, 8, 2, 3, 5, 7, 0, 1, 10]`

  Output: [4, 9]

</details>

#### Starter Code for Question 3


```python
from typing import List

def missing_num(nums: List) -> int:
  if not nums:
        return -1

  max_val = max(nums)
  present = set(nums)

  missing = [i for i in range(max_val + 1) if i not in present]

  return missing if missing else -1

print(missing_num([0, 2])) # [1]
print(missing_num([5, 0, 1]))  # [2, 3, 4]
print(missing_num([6, 8, 2, 3, 5, 7, 0, 1, 10]))  # [4, 9]
print(missing_num([0, 1, 2, 3])) # -1
```


-   Copy the solution your partner wrote. 


```python
# Your answer here
s = "{[(]}"  
open_brackets = ["(", "[", "{"]
close_brackets = [")", "]", "}"]

def is_valid_brackets(s: str) -> bool:
    if len(s) % 2 != 0:
        return False
    elif s[0] in close_brackets:
        return False
    
    bracket_stack = []

    for bracket in s:
        if bracket in open_brackets:
            bracket_stack.append(bracket)
        elif bracket in close_brackets:
            if not bracket_stack:
                return False
            elif close_brackets.index(bracket) == open_brackets.index(bracket_stack[-1]):
                bracket_stack.pop()
            else:
                return False
    
    if len(bracket_stack) == 0:
            return True
    else:
         return False
                     

is_valid_brackets(s)
```
